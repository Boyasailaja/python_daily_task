```python
# food management

class Food(ABC):
    
    # constructor :
    def __init__(self,f_name,f_price):
        
        self.__Food_name = f_name
        self.__Food_price = f_price
        
    # getter name 
    def get_food_name(self):
        return self.__Food_name
    
    # getter price
    def get_food_price(self):
        return self.__Food_price
    
    # abstract method 
    @abstractmethod
    def prepare(self):
        pass
    
    # abstract method:
    @abstractmethod 
    def deliver(self):
        pass
    
    # display all details 
    def display_all_details(self):
        print(f"Food Name  : {self.__Food_name}")
        print(f"Food Price  : {self.__Food_price}")
        self.prepare()
        self.deliver()
        
    
    
# child class
class Pizza(Food):
    
    def prepare(self):
        print('Pizza is prepared ')
        
    def deliver(self):
        print("Pizza is Delivered")
        
# child class 2:
class Burger(Food):
    
    def prepare(self):
        print('Burger is prepared ')
        
    def deliver(self):
        print("Burger is Delivered") 
        
# child class 3:
class Biryani(Food):
    
    def prepare(self):
        print('Biryani is prepared ')
        
    def deliver(self):
        print("Biryani is Delivered") 
        
# create the class object

pizza = Pizza("Pizza",500)
burger = Burger("Burger",200)
birynai = Biryani("Biryani",700)


birynai.display_all_details()
```

```python
# payments  
from abc import ABC , abstractmethod

class Payment(ABC):
    
    # constructor :
    def __init__(self,owner,amount):
        self.__Owner_name = owner  # private 
        self.__Amount  = amount    # private
        
        
    # getter : get the owner name 
    def get_owner(self):
        return f"Owner Name : {self.__Owner_name}"
    
    # getter : get the Amount 
    def get_amount(self):
        return f"Amount : {self.__Amount}"
    
    #setter 
    
    def set_amount(self,amt):
        
        if amt > 0:
            self.__Amount += amt
            return f"Amount updated : {self.__Amount}"
        
        else :
            print("Please enter the amount more than zero")
            
    # abstract method :
    
    @abstractmethod
    def pay(self):
        pass 
    
    # instace method 
    def recipt(self):
        return f"Owner Name : {self.__Owner_name} \nAmount : {self.__Amount}"
        
    
    
    
# child class 
class Credit_card(Payment):
    
    # constructor
    def __init__(self,owner,amount,credit_card_no):
        super().__init__(owner,amount)
        self.__Credit_card =credit_card_no
        
        
    def pay(self):
        print("Credit card details :")
        print(f"Credit Card Number  : {self.__Credit_card}")
        print(self.recipt())
        
# child class 
class Debit_card(Payment):
    
    # constructor
    def __init__(self,owner,amount,debit_card_no):
        super().__init__(owner,amount)
        self.__Debit_card =debit_card_no
        
        
    def pay(self):
        print("Debit card details :")
        print(f"Debit Card Number  : {self.__Debit_card}")
        print(self.recipt())
        
        
# child class 
class UPI(Payment):
    
    # constructor
    def __init__(self,owner,amount,upi_id):
        super().__init__(owner,amount)
        self.__UPI_id =upi_id
        
        
    def pay(self):
        print("UIP payments details :")
        print(f"UPI  : {self.__UPI_id}")
        print(self.recipt())


        
# create the class object

credit  = Credit_card("Sumit",1000,12345)
debit   = Debit_card ("Parbhas",2000,35622)
upi     = UPI("Rocky Bhai",5000,89755)

# call the credit method and attributes

#print(credit.get_owner())
#print(credit.get_amount())
#print(credit.recipt())
#print(credit.pay())
   
# call the credit method and attributes

#print(debit.get_owner())
#print(debit.get_amount())
#print(credit.recipt())
#print(debit.pay())
    
    
# call the UPI method and attributes

#print(upi.get_owner())
#print(upi.get_amount())
#print(upi.recipt())
print(upi.pay())   
    
    
   
        
```

```python
# Grand Parent --parent  - child

# grand Parent
class Person:
    
    # constructor:
    def __init__(self,name,age):
        self.__Name = name  # private
        self.__Age = age   # private
        
        
    # getter 
    def get_name(self):
        return f"Name of Person : {self.__Name}"
    
    # getter
    def get_age(self):
        return f"Age of Person : {self.__Age}"
    
    
    # setter of age
    
    def set_age(self,age):
        if 20 < age < 60:
            self.__Age = age
            print(f"Age is Updated : {self.__Age} ")
            
        else:
            print("Please enter the age between the 20 and 60")
            
            
        
    # method 
    def person_info(self):
        return f"Name of Person : {self.__Name} \nAge of Person : {self.__Age}"
    
    
# Parent class :

class Employee(Person):
    
    # constructor 
    def __init__(self,name,age,salary,company):
        super().__init__(name,age)
        self.__Employee_Salary = salary
        self.__Company_name = company
        
        
        
    # getter 
    def get_salary(self):
        return f"Salary of Employee : {self.__Employee_Salary}"
    
    # getter
    def get_company(self):
        return f"Company name  of Employee : {self.__Company_name}"  
    
    
    # setter 
    def set_salary(self,sal):
        
        if sal <= (20/100) * self.__Employee_Salary:
            self.__Employee_Salary += sal
            print(f"Salary is updated : {sal}")
            
        else:
            print("Please updated the salary less than  20% of cuurent salary")
            
            
    def employee_info(self):
        print(f"Employee Salary :{self.__Employee_Salary}")
        print(f"Employee Company Name : {self.__Company_name}")
        print(self.person_info())
        
    
# child class
class Manager(Employee):
    
    # constructor 
    def __init__(self,name,age,salary,company,team_size,bonus):
        super().__init__(name,age,salary,company)
        self.__Team_size = team_size
        self.__Bonus = bonus
        
        
        
    # getter 
    def get_teamsize(self):
        return f"Team Size : {self.__Team_size}"
    
    # getter
    def get_bonus(self):
        return f"Bonus of Employee : {self.__Bonus}" 
    
    
    # setter 
    
    def set_team(self,new_team):
        
        if 5 < new_team < 10:
            self.__Team_size += new_team
            return f"Team Is updated : {self.__Team_size}"
        
        else:
            print("please enter the team size inbetween 5 and 10")
            
            
    # setter
    
    def set_bonus(self,new_bonus):
        
        if new_bonus <= (10/100) * self.__Employee_Salary:
            self.__Bonus = new_bonus
            return f"Update bonus {new_bonus}"
        
        
    def displat_all_info(self):
        print(f"Team Size : {self.__Team_size}")
        print(f"Bonus  : {self.__Bonus}")
        print(self.employee_info())  
# create the object 

manager = Manager("Rajesh",45,40000,"Google",12,3000)
manager.displat_all_info()
```

```python
#6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict. 

class ClassA:
    def display(self):
        print("This is Class A")

class ClassB:
    def display(self):
        print("This is Class B")

class ChildClass(ClassA, ClassB):
    def display(self):
        ClassA.display(self)  # Call ClassA's display method
        ClassB.display(self)  # Call ClassB's display method

# Create an instance of ChildClass and call the display method
child = ChildClass()
child.display()
```

### Type 4 — Hierarchical Inheritance

```python
# Parent  : Employee
# attribute :name , salary,dept all private
# getter and setter : use static method for set_sal :sal > 0
# class atribute : total_emp = 0
# class method : get_total()
# method : work() -- print working meassge
# method calculate_bonus - return of 5% of sal
# method show == print all +bonus

# create the 3 child 
# child 1 : devloper : add langauge, bonus 20 %
# overide work () --writing code 

# child 1 : Designer : add tool, bonus 15 %
# overide work () --Creating UI

# child 1 : manager : add team_size, bonus 20 %
# overide work () --Managing Team

class Employee:
    total_emp = 0  # class attribute
    
    def __init__(self, name, salary, dept):
        # private attributes
        self.__name = name
        self.__salary = salary
        self.__dept = dept
        Employee.total_emp += 1  # increase count
    
    # Getters
    def get_name(self):
        return self.__name
    
    def get_salary(self):
        return self.__salary
    
    def get_dept(self):
        return self.__dept
    
    # Setter with static method
    @staticmethod
    def set_sal(sal):
        if sal > 0:
            return sal
        else:
            print("Salary must be greater than 0")
            return 0
    
    def set_salary(self, sal):
        self.__salary = Employee.set_sal(sal)
    
    @classmethod
    def get_total(cls):
        return f"Total Employees: {cls.total_emp}"
    
    def work(self):
        print("Employee is working")
    
    def calculate_bonus(self):
        return 0.05 * self.__salary  # 5% bonus
    
    def show(self):
        bonus = self.calculate_bonus()
        print(f"Name: {self.__name}")
        print(f"Dept: {self.__dept}")
        print(f"Salary: {self.__salary}")
        print(f"Bonus: {bonus}")
        print("-" * 20)

# Child 1: Developer
class Developer(Employee):
    def __init__(self, name, salary, dept, language):
        super().__init__(name, salary, dept)
        self.language = language
    
    def work(self):  # override
        print("Writing Code")
    
    def calculate_bonus(self):  # override - 20%
        return 0.20 * self.get_salary()
    
    def show(self):
        super().show()
        print(f"Language: {self.language}")

# Child 2: Designer
class Designer(Employee):
    def __init__(self, name, salary, dept, tool):
        super().__init__(name, salary, dept)
        self.tool = tool
    
    def work(self):  # override
        print("Creating UI")
    
    def calculate_bonus(self):  # override - 15%
        return 0.15 * self.get_salary()
    
    def show(self):
        super().show()
        print(f"Tool: {self.tool}")

# Child 3: Manager
class Manager(Employee):
    def __init__(self, name, salary, dept, team_size):
        super().__init__(name, salary, dept)
        self.team_size = team_size
    
    def work(self):  # override
        print("Managing Team")
    
    def calculate_bonus(self):  # override - 20%
        return 0.20 * self.get_salary()
    
    def show(self):
        super().show()
        print(f"Team Size: {self.team_size}")

# Testing
d1 = Developer("Aarav", 50000, "IT", "Python")
des1 = Designer("Pooja", 40000, "Design", "Figma")
m1 = Manager("Rahul", 80000, "HR", 12)

d1.work()
d1.show()

des1.work()
des1.show()

m1.work()
m1.show()

print(Employee.get_total())
```
