```python
# Pyplot Common Commands 
import matplotlib.pyplot as plt
import numpy as np
X = np.linspace(0,10,100)

y = np.sin(X)


# create the plot
plt.plot(X,y,
         color = "Red",
        linewidth=0.5,
        marker = "*",
        markersize = 8,
        markeredgecolor='yellow',
        markerfacecolor='pink',
        )

# commands
plt.title("Sine Waves")
plt.xlabel("X Axis Data ")
plt.ylabel("Sin X Data ")
plt.xlim(1,10)      # xrange
plt.ylim(-1.5, 1.5)  # yrange
plt.xticks([0,2,4,6,8,10])      # set x tick marks
plt.yticks([-1, 0, 1])          # set y tick marks 
plt.grid(True, alpha=0.9)       # show grid 
plt.legend(["Sin X Waves"])          # add legend 
plt.tight_layout()              # auto-fix spacing
plt.show()
```

```python
# Multiplot Figure

X =np.linspace(0,10,100)

#create the Multiplot figure

fig,axes = plt.subplots(nrows = 2,ncols = 2,
                        dpi=100,
                       figsize = (12,8))


# figure method
fig.set_size_inches(5,4)   # resize figure in inches
fig.set_facecolor("pink")  # change background color
fig.suptitle("Figure Title",
              fontsize=6,  # font size figure title
             fontweight='bold')

# Plot in each subplot
axes[0,0].plot(X, np.sin(X), "olive")
axes[0,0].set_title("Sine Plot")

# Plot in each subplot
axes[0,1].plot(X, np.cos(X), "slategray")
axes[0,1].set_title("Cos Plot")

# Plot in each subplot
axes[1,0].plot(X, np.tanh(X), "blue")
axes[1,0].set_title("Tanh Plot")

# Plot in each subplot
axes[1,1].plot(X, X**2, "tan")
axes[1,1].set_title("Squre Plot")


plt.show
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\yuvra\\Downloads\\matplotlib_sales_clean_1000.csv")

df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Check columns
print(df.columns)

# Convert Date column if required
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Create Month column
df["Month"] = df["Order_Date"].dt.month_name()

# Month order
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

df["Month"] = pd.Categorical(df["Month"], categories=month_order, ordered=True)
```

# Monthly Revenue Line Chart

```python
monthly_revenue = df.groupby("Month", observed=True)["Revenue"].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly_revenue.index, monthly_revenue.values,
         marker="o", label="Revenue")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

# Monthly Revenue and Profit

```python
monthly = df.groupby("Month", observed=True)[["Revenue", "Profit"]].sum()

plt.figure(figsize=(10,5))

plt.plot(monthly.index, monthly["Revenue"],
         marker="o", linestyle="-", label="Revenue")

plt.plot(monthly.index, monthly["Profit"],
         marker="s", linestyle="--", label="Profit")

plt.title("Monthly Revenue and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

plt.show()
```

# Revenue by Product Category

```python
category_revenue = df.groupby("Product_Category")["Revenue"].sum()
category_revenue = category_revenue.sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(category_revenue.index, category_revenue.values)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.grid(axis="y")

plt.show()
```

# Profit by City and Highlight Highest

```python
city_profit = df.groupby("City")["Profit"].sum()
highest_city = city_profit.idxmax()

colors = ["gray"] * len(city_profit)
colors[list(city_profit.index).index(highest_city)] = "red"

plt.figure(figsize=(8,5))
plt.bar(city_profit.index, city_profit.values, color=colors)

plt.title("Profit by City")
plt.xlabel("City")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.grid(axis="y")

plt.annotate(
    "Highest Profit",
    xy=(highest_city, city_profit[highest_city]),
    xytext=(0,20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

plt.show()
```

# Top 10 Orders Based on Revenue

```python
top10 = df.nlargest(10, "Revenue")

plt.figure(figsize=(10,6))
plt.barh(top10["Order_ID"].astype(str), top10["Revenue"])

plt.title("Top 10 Orders by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Order ID")
plt.gca().invert_yaxis()

plt.show()
```

# 3-Month Moving Average 

```python
monthly_revenue = df.groupby("Month", observed=True)["Revenue"].sum()

moving_avg = monthly_revenue.rolling(3).mean()

plt.figure(figsize=(10,5))

plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o",
    label="Revenue"
)

plt.plot(
    monthly_revenue.index,
    moving_avg.values,
    marker="s",
    linestyle="--",
    label="3-Month Moving Average"
)

plt.title("Monthly Revenue with 3-Month Moving Average")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

plt.show()
```

# Correlation Heatmap Using Matplotlib Only

```python
columns = [
    "Customer_Age",
    "Units_Sold",
    "Unit_Price",
    "Discount_Pct",
    "Revenue",
    "Cost",
    "Profit"
]

corr = df[columns].corr()

plt.figure(figsize=(10,8))

plt.imshow(corr, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(columns)),
    columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(columns)),
    columns
)

for i in range(len(columns)):
    for j in range(len(columns)):
        plt.text(
            j, i,
            f"{corr.iloc[i,j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```
