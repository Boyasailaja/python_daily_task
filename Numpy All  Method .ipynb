```python
# id  : 100
# Name 
# Gender
# Dept
# Exp
# Role 
# Salary


# ID 
ID =  np.arange(100,200)



# name :

name = ["Sumit","Vaishnavi","Ayushi","Sameer","Sahil","Neha","Rahul","Danish","Sitara","Haritha","Akshitha"]

Name = np.random.choice(name,100)


# gender

g = ["Male","Female"]
Gender =  np.random.choice(g,100)

# dept

dept= ["IT","Finace","HR","Inventory","Call Center","Science"]

Dept = np.random.choice(dept,100)

# Exp 

exp = np.random.uniform(1,20,100)
exp = np.round(exp,1)
exp

# role

role = ["Devloper","Testing","Data Scientist","AI Engineer","Devops"]

Role = np.random.choice(role,100)

# salary

salary = np.random.uniform(45000, 100000,100)
salary = np.round(salary,2)
 
len(salary)
```

```python
import pandas as pd
df = pd.DataFrame({
    "ID":ID,
    "Name":Name,
    "Gender":Gender,
    "Department":Dept,
    "Exp":exp,
    "Job Role":Role,
    "Salary":salary
})
```
