```python
from abc import ABC, abstractmethod


# Custom Exceptions
class InsufficientBalanceError(Exception):
    pass

class InvalidAmountError(Exception):
    pass

class AccountNotFoundError(Exception):
    pass

class DailyLimitExceededError(Exception):
    pass


# Abstract Class
class Account(ABC):

    total_accounts = 0
    next_account_no = 1001

    def __init__(self, name):
        self.__account_no = Account.next_account_no
        Account.next_account_no += 1
        Account.total_accounts += 1

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def get_balance(self):
        pass

    def show_account_info(self):
        print("Account No:", self.__account_no)

    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)


# Savings Account
class SavingsAccount(Account):

    daily_limit = 25000
    minimum_balance = 1000
    interest_rate = 4

    def __init__(self, name):
        super().__init__(name)
        self.__account_no = Account.next_account_no - 1
        self.__name = name
        self.__balance = 0
        self.__daily_withdrawn = 0

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount < 0:
            raise InvalidAmountError("Balance cannot be negative")
        self.__balance = amount

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0")

        self.__balance += amount
        print(f"₹{amount} deposited successfully")

    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than 0")

        if amount > self.__balance:
            raise InsufficientBalanceError("Insufficient Balance")

        if self.__daily_withdrawn + amount > self.daily_limit:
            raise DailyLimitExceededError("Daily withdrawal limit is ₹25000")

        if self.__balance - amount < self.minimum_balance:
            raise InsufficientBalanceError("Minimum balance of ₹1000 must be maintained")

        self.__balance -= amount
        self.__daily_withdrawn += amount

        print(f"₹{amount} withdrawn successfully")

    def show_account_info(self):
        super().show_account_info()
        print("Name:", self.__name)
        print("Account Type: Savings")
        print("Balance:", self.__balance)
        print("Interest Rate:", self.interest_rate, "%")


# Current Account
class CurrentAccount(Account):

    overdraft_limit = 10000

    def __init__(self, name):
        super().__init__(name)
        self.__account_no = Account.next_account_no - 1
        self.__name = name
        self.__balance = 0

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount < 0:
            raise InvalidAmountError("Balance cannot be negative")

        self.__balance = amount

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0")

        self.__balance += amount
        print(f"₹{amount} deposited successfully")

    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than 0")

        if amount > self.__balance + self.overdraft_limit:
            raise InsufficientBalanceError("Overdraft limit exceeded")

        self.__balance -= amount
        print(f"₹{amount} withdrawn successfully")

    def show_account_info(self):
        super().show_account_info()
        print("Name:", self.__name)
        print("Account Type: Current")
        print("Balance:", self.__balance)
        print("Overdraft Limit:", self.overdraft_limit)


# Bank Class
class Bank:

    def __init__(self):
        self.accounts = {}

    def create_account(self):

        print("1. Savings Account")
        print("2. Current Account")

        choice = int(input("Enter type: "))
        name = input("Enter Name: ")

        if choice == 1:
            account = SavingsAccount(name)

        elif choice == 2:
            account = CurrentAccount(name)

        else:
            print("Invalid Account Type")
            return

        acc_no = Account.next_account_no - 1
        self.accounts[acc_no] = account

        print("Account Created Successfully!")
        print("Account No:", acc_no)


    def find_account(self, acc_no):

        if acc_no not in self.accounts:
            raise AccountNotFoundError("Account does not exist")

        return self.accounts[acc_no]


    def deposit(self):

        try:
            acc_no = int(input("Enter Account No: "))
            amount = float(input("Enter Amount: "))

            account = self.find_account(acc_no)
            account.deposit(amount)

        except Exception as e:
            print("Error:", type(e).__name__, "-", e)


    def withdraw(self):

        try:
            acc_no = int(input("Enter Account No: "))
            amount = float(input("Enter Amount: "))

            account = self.find_account(acc_no)
            account.withdraw(amount)

        except Exception as e:
            print("Error:", type(e).__name__, "-", e)


    def check_balance(self):

        try:
            acc_no = int(input("Enter Account No: "))

            account = self.find_account(acc_no)

            print("Balance:", account.get_balance())

        except Exception as e:
            print("Error:", type(e).__name__, "-", e)


    def show_info(self):

        try:
            acc_no = int(input("Enter Account No: "))

            account = self.find_account(acc_no)

            account.show_account_info()

        except Exception as e:
            print("Error:", type(e).__name__, "-", e)


# Main Program
bank = Bank()

while True:

    print("\n===== Welcome to SBI Bank =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Show Account Info")
    print("6. Total Accounts")
    print("7. Exit")

    try:

        choice = int(input("\nEnter choice: "))

        if choice == 1:
            bank.create_account()

        elif choice == 2:
            bank.deposit()

        elif choice == 3:
            bank.withdraw()

        elif choice == 4:
            bank.check_balance()

        elif choice == 5:
            bank.show_info()

        elif choice == 6:
            Account.show_total_accounts()

        elif choice == 7:
            print("Thank you for using XYZ Bank")
            break

        else:
            print("Invalid Choice")

    except ValueError:
        print("Please enter numbers only")
```
