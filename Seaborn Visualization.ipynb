```python
# ============================================================
# MATPLOTLIB + SEABORN - HOTEL BOOKING DATASET
# 20 QUESTIONS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# LOAD DATASET
# ------------------------------------------------------------
# If your dataset is already loaded, skip this line.
df = pd.read_csv("C:\\Users\\yuvra\\Downloads\\Hotel_booking_dataset.csv")

# Display columns
print(df.columns)


# ------------------------------------------------------------
# CLEAN COLUMN NAMES
# ------------------------------------------------------------
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("%", "percent")
)


# ------------------------------------------------------------
# BASIC DATA CLEANING
# ------------------------------------------------------------

# Age
df["age"] = (
    df["age"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Remove impossible ages
df.loc[(df["age"] < 1) | (df["age"] > 100), "age"] = np.nan


# Room Rate
df["room_rate"] = (
    df["room_rate"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["room_rate"] = pd.to_numeric(df["room_rate"], errors="coerce")


# Final Amount
df["final_amount"] = (
    df["final_amount"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["final_amount"] = pd.to_numeric(df["final_amount"], errors="coerce")


# Room Nights
df["room_nights"] = (
    df["room_nights"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["room_nights"] = pd.to_numeric(
    df["room_nights"],
    errors="coerce"
)


# Discount %
df["discount_percent"] = (
    df["discount_percent"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["discount_percent"] = pd.to_numeric(
    df["discount_percent"],
    errors="coerce"
)


# Rating
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)

# Keep only ratings from 1 to 5
df.loc[
    ~df["rating"].between(1, 5),
    "rating"
] = np.nan


# Guests
df["guests"] = pd.to_numeric(
    df["guests"],
    errors="coerce"
)


# Profit
df["profit"] = (
    df["profit"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(-?\d+\.?\d*)")[0]
)
df["profit"] = pd.to_numeric(
    df["profit"],
    errors="coerce"
)


# Gross Amount
df["gross_amount"] = (
    df["gross_amount"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
)
df["gross_amount"] = pd.to_numeric(
    df["gross_amount"],
    errors="coerce"
)


# ------------------------------------------------------------
# STANDARDIZE CATEGORICAL COLUMNS
# ------------------------------------------------------------

# Gender
df["gender"] = (
    df["gender"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female",
        "other": "Other"
    })
)


# Room Type
df["room_type"] = (
    df["room_type"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Booking Channel
df["booking_channel"] = (
    df["booking_channel"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Customer Segment
df["customer_segment"] = (
    df["customer_segment"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Payment Method
df["payment_method"] = (
    df["payment_method"]
    .astype(str)
    .str.strip()
    .str.title()
)


# City
df["city"] = (
    df["city"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Is Member
df["is_member"] = (
    df["is_member"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "yes": "Yes",
        "y": "Yes",
        "1": "Yes",
        "true": "Yes",
        "no": "No",
        "n": "No",
        "0": "No",
        "false": "No"
    })
)


# Cancelled
df["cancelled"] = (
    df["cancelled"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "yes": 1,
        "y": 1,
        "true": 1,
        "1": 1,
        "no": 0,
        "n": 0,
        "false": 0,
        "0": 0
    })
)

df["cancelled"] = pd.to_numeric(
    df["cancelled"],
    errors="coerce"
)


# ============================================================
# Q1. HISTOGRAM OF AGE
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df["age"].dropna(),
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Customer Ages")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()


# ============================================================
# Q2. COUNT PLOT OF ROOM TYPE
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="room_type"
)

plt.title("Number of Bookings by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q3. BAR PLOT OF AVERAGE FINAL AMOUNT BY CITY
# ============================================================

city_avg = (
    df.groupby("city")["final_amount"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

city_avg.plot(kind="bar")

plt.title("Average Final Amount by City")
plt.xlabel("City")
plt.ylabel("Average Final Amount")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# Q4. BOX PLOT OF ROOM RATE BY ROOM TYPE
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="room_type",
    y="room_rate"
)

plt.title("Room Rate by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Room Rate")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q5. COUNT PLOT OF BOOKING CHANNEL
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="booking_channel"
)

plt.title("Bookings by Booking Channel")
plt.xlabel("Booking Channel")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q6. HISTOGRAM OF ROOM NIGHTS
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df["room_nights"].dropna(),
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Room Nights")
plt.xlabel("Room Nights")
plt.ylabel("Number of Bookings")

plt.show()


# ============================================================
# Q7. BAR PLOT OF CANCELLATION RATE BY CUSTOMER SEGMENT
# ============================================================

cancel_rate = (
    df.groupby("customer_segment")["cancelled"]
    .mean() * 100
)

plt.figure(figsize=(8, 5))

cancel_rate.plot(kind="bar")

plt.title("Cancellation Rate by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Cancellation Rate (%)")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q8. BOX PLOT OF AGE BY GENDER
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="gender",
    y="age"
)

plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")

plt.show()


# ============================================================
# Q9. COUNT PLOT OF PAYMENT METHOD
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="payment_method"
)

plt.title("Popularity of Payment Methods")
plt.xlabel("Payment Method")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q10. SCATTER PLOT OF ROOM RATE VS FINAL AMOUNT
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="room_rate",
    y="final_amount"
)

plt.title("Room Rate vs Final Amount")
plt.xlabel("Room Rate")
plt.ylabel("Final Amount")

plt.show()


# ============================================================
# Q11. VIOLIN PLOT OF FINAL AMOUNT BY ROOM TYPE
# ============================================================

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="room_type",
    y="final_amount"
)

plt.title("Final Amount Distribution by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Final Amount")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q12. BAR PLOT OF AVERAGE DISCOUNT % BY BOOKING CHANNEL
# ============================================================

discount_avg = (
    df.groupby("booking_channel")["discount_percent"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

discount_avg.plot(kind="bar")

plt.title("Average Discount % by Booking Channel")
plt.xlabel("Booking Channel")
plt.ylabel("Average Discount (%)")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q13. HISTOGRAM OF RATING
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df["rating"].dropna(),
    bins=5,
    edgecolor="black"
)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")

plt.xticks([1, 2, 3, 4, 5])

plt.show()


# ============================================================
# Q14. COUNT PLOT OF IS MEMBER
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="is_member"
)

plt.title("Members vs Non-Members")
plt.xlabel("Membership Status")
plt.ylabel("Number of Customers")

plt.show()


# ============================================================
# Q15. BOX PLOT OF PROFIT BY CITY
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="city",
    y="profit"
)

plt.title("Profit Distribution by City")
plt.xlabel("City")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# Q16. BAR PLOT OF AVERAGE GUESTS BY ROOM TYPE
# ============================================================

guest_avg = (
    df.groupby("room_type")["guests"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

guest_avg.plot(kind="bar")

plt.title("Average Guests by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Number of Guests")

plt.xticks(rotation=30)

plt.show()


# ============================================================
# Q17. KDE PLOT OF AGE
# ============================================================

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x="age",
    fill=True
)

plt.title("Age Density Distribution")
plt.xlabel("Age")
plt.ylabel("Density")

plt.show()


# ============================================================
# Q18. HEATMAP OF CORRELATION
# ============================================================

numeric_columns = [
    "age",
    "room_rate",
    "final_amount",
    "profit",
    "discount_percent"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numeric Variables")

plt.show()


# ============================================================
# Q19. COUNT PLOT OF GENDER WITH CUSTOMER SEGMENT
# ============================================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="gender",
    hue="customer_segment"
)

plt.title("Gender Distribution by Customer Segment")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.show()


# ============================================================
# Q20. BAR PLOT OF TOTAL GROSS AMOUNT BY MONTH
# ============================================================

# Convert Booking Date to datetime

df["booking_date"] = pd.to_datetime(
    df["booking_date"],
    errors="coerce"
)

# Extract month

df["month"] = df["booking_date"].dt.month_name()

# Calculate total gross amount

monthly_revenue = (
    df.groupby("month")["gross_amount"]
    .sum()
)

# Correct month order

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

monthly_revenue = monthly_revenue.reindex(
    month_order
)

plt.figure(figsize=(12, 5))

monthly_revenue.plot(kind="bar")

plt.title("Total Gross Amount by Month")
plt.xlabel("Month")
plt.ylabel("Total Gross Amount")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# END OF 20 QUESTIONS
# ============================================================
```

```python
# ============================================================
# HOTEL DATASET - 30 DATA CLEANING QUESTIONS
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv("C:\\Users\\yuvra\\Downloads\\Hotel_booking_dataset.csv")

print("Original Shape:", df.shape)
print(df.head())


# ============================================================
# Q1. REMOVE LEADING/TRAILING SPACES FROM COLUMN NAMES
# ============================================================

df.columns = df.columns.str.strip()

print("\nClean Column Names:")
print(df.columns.tolist())


# ============================================================
# Q2. STANDARDIZE ALL DATE FORMATS
# ============================================================

date_columns = ["Booking Date"]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col],
            errors="coerce",
            dayfirst=True
        )
        df[col] = df[col].dt.strftime("%Y-%m-%d")

print("\nBooking Date:")
print(df["Booking Date"].head())


# ============================================================
# Q3. HANDLE COMPLETELY MISSING CUSTOMER NAME VALUES
# ============================================================

df["Customer Name"] = (
    df["Customer Name"]
    .astype(str)
    .str.strip()
)

df["Customer Name"] = df["Customer Name"].replace(
    ["", "N/A", "NA", "na", "None", "nan"],
    np.nan
)

df["Customer Name"] = df["Customer Name"].fillna("Unknown")


# ============================================================
# Q4. FIX INCONSISTENT GENDER VALUES
# ============================================================

df["Gender"] = (
    df["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Gender"] = df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "other": "Other"
})

df.loc[
    ~df["Gender"].isin(["Male", "Female", "Other"]),
    "Gender"
] = "Other"


# ============================================================
# Q5. CLEAN AND STANDARDIZE CITY NAMES
# ============================================================

df["City"] = (
    df["City"]
    .astype(str)
    .str.strip()
    .str.title()
)

df["City"] = df["City"].replace({
    "Hydrabad": "Hyderabad",
    "Bangalore": "Bengaluru",
    "Bengaluru": "Bengaluru",
    "Mumbai": "Mumbai",
    "Hyderabad": "Hyderabad"
})

df["City"] = df["City"].replace(
    ["", "Nan", "N/A", "Na"],
    "Unknown"
)


# ============================================================
# Q6. STANDARDIZE ROOM TYPE
# ============================================================

df["Room Type"] = (
    df["Room Type"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Room Type"] = df["Room Type"].replace({
    "standard": "Standard",
    "std": "Standard",
    "deluxe": "Deluxe",
    "delux": "Deluxe",
    "suite": "Suite",
    "exec": "Executive",
    "executive": "Executive"
})

df.loc[
    ~df["Room Type"].isin(
        ["Standard", "Deluxe", "Suite", "Executive"]
    ),
    "Room Type"
] = "Unknown"


# ============================================================
# Q7. CLEAN BOOKING CHANNEL
# ============================================================

df["Booking Channel"] = (
    df["Booking Channel"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Booking Channel"] = df["Booking Channel"].replace({
    "online": "Online",
    "walk-in": "Walk-in",
    "walkin": "Walk-in",
    "corporate": "Corporate",
    "corp": "Corporate",
    "travel agent": "Travel Agent",
    "travelagent": "Travel Agent"
})

df.loc[
    ~df["Booking Channel"].isin(
        ["Online", "Walk-in", "Corporate", "Travel Agent"]
    ),
    "Booking Channel"
] = "Unknown"


# ============================================================
# Q8. STANDARDIZE CUSTOMER SEGMENT
# ============================================================

df["Customer Segment"] = (
    df["Customer Segment"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Customer Segment"] = df["Customer Segment"].replace({
    "premium": "Premium",
    "regular": "Regular",
    "reg": "Regular",
    "budget": "Budget",
    "corporate": "Corporate",
    "corp": "Corporate",
    "vip": "VIP"
})

df.loc[
    ~df["Customer Segment"].isin(
        ["Budget", "Regular", "Premium", "Corporate", "VIP"]
    ),
    "Customer Segment"
] = "Unknown"


# ============================================================
# Q9. FIX AGE COLUMN
# ============================================================

df["Age"] = (
    df["Age"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Age below 18 or above 100 = missing
df.loc[
    (df["Age"] < 18) | (df["Age"] > 100),
    "Age"
] = np.nan

df["Age"] = df["Age"].round().astype("Int64")


# ============================================================
# Q10. CLEAN ROOM NIGHTS
# ============================================================

df["Room Nights"] = (
    df["Room Nights"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Room Nights"] = pd.to_numeric(
    df["Room Nights"],
    errors="coerce"
)

# Extremely high values are treated as missing
df.loc[
    df["Room Nights"] > 30,
    "Room Nights"
] = np.nan

df["Room Nights"] = df["Room Nights"].round().astype("Int64")


# ============================================================
# Q11. FIX ROOM RATE
# ============================================================

df["Room Rate"] = (
    df["Room Rate"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Room Rate"] = pd.to_numeric(
    df["Room Rate"],
    errors="coerce"
)

# Flag unrealistic rates
df.loc[
    df["Room Rate"] > 100000,
    "Room Rate"
] = np.nan


# ============================================================
# Q12. CLEAN DISCOUNT %
# ============================================================

df["Discount %"] = (
    df["Discount %"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Discount %"] = df["Discount %"].replace(
    ["none", "n/a", "na", ""],
    "0"
)

df["Discount %"] = (
    df["Discount %"]
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Discount %"] = pd.to_numeric(
    df["Discount %"],
    errors="coerce"
)

# Values greater than 100 are invalid
df.loc[
    df["Discount %"] > 100,
    "Discount %"
] = np.nan


# ============================================================
# Q13. HANDLE RATING VALUES
# ============================================================

df["Rating"] = (
    df["Rating"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Rating"] = pd.to_numeric(
    df["Rating"],
    errors="coerce"
)

# Keep only ratings 1 to 5
df.loc[
    ~df["Rating"].between(1, 5),
    "Rating"
] = np.nan


# ============================================================
# Q14. CLEAN GUESTS
# ============================================================

df["Guests"] = pd.to_numeric(
    df["Guests"],
    errors="coerce"
)

# More than 10 guests considered extreme
df.loc[
    df["Guests"] > 10,
    "Guests"
] = np.nan

df["Guests"] = df["Guests"].round().astype("Int64")


# ============================================================
# Q15. STANDARDIZE IS MEMBER
# ============================================================

df["Is Member"] = (
    df["Is Member"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Is Member"] = df["Is Member"].replace({
    "1": "Yes",
    "y": "Yes",
    "yes": "Yes",
    "member": "Yes",
    "0": "No",
    "n": "No",
    "no": "No"
})

df.loc[
    ~df["Is Member"].isin(["Yes", "No"]),
    "Is Member"
] = "No"


# ============================================================
# Q16. STANDARDIZE CANCELLED
# ============================================================

df["Cancelled"] = (
    df["Cancelled"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Cancelled"] = df["Cancelled"].replace({
    "1": 1,
    "yes": 1,
    "y": 1,
    "true": 1,
    "0": 0,
    "no": 0,
    "n": 0,
    "false": 0
})

df["Cancelled"] = pd.to_numeric(
    df["Cancelled"],
    errors="coerce"
)


# ============================================================
# Q17. CLEAN PAYMENT METHOD
# ============================================================

df["Payment Method"] = (
    df["Payment Method"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Payment Method"] = df["Payment Method"].replace({
    "credit card": "Credit Card",
    "creditcard": "Credit Card",
    "debit card": "Debit Card",
    "debitcard": "Debit Card",
    "upi": "UPI",
    "net banking": "Net Banking",
    "netbanking": "Net Banking",
    "cash": "Cash"
})

df.loc[
    ~df["Payment Method"].isin(
        [
            "Credit Card",
            "Debit Card",
            "UPI",
            "Net Banking",
            "Cash"
        ]
    ),
    "Payment Method"
] = "Unknown"


# ============================================================
# Q18. REMOVE DUPLICATE BOOKING IDs
# ============================================================

print("\nDuplicate Booking IDs:")
print(df["Booking ID"].duplicated().sum())

# Keep first occurrence
df = df.drop_duplicates(
    subset="Booking ID",
    keep="first"
)


# ============================================================
# Q19. HANDLE MISSING ROOM TYPE
# ============================================================

df["Room Type"] = df["Room Type"].replace(
    ["", "nan", "NaN"],
    np.nan
)

df["Room Type"] = df["Room Type"].fillna(
    "Unknown"
)


# ============================================================
# Q20. FIX FINAL AMOUNT CALCULATIONS
# ============================================================

# Convert financial columns to numeric first

financial_columns = [
    "Gross Amount",
    "Discount Amount",
    "Final Amount"
]

for col in financial_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.extract(r"(-?\d+\.?\d*)")[0]
    )

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# Calculate expected Final Amount

df["Expected Final Amount"] = (
    df["Gross Amount"] -
    df["Discount Amount"]
)

# Difference between actual and expected

df["Final Amount Difference"] = (
    df["Final Amount"] -
    df["Expected Final Amount"]
)

# Flag calculation errors

df["Final Amount Check"] = np.where(
    abs(df["Final Amount Difference"]) > 1,
    "Check",
    "Correct"
)


# ============================================================
# Q21. CLEAN CUSTOMER NAME
# ============================================================

df["Customer Name"] = (
    df["Customer Name"]
    .astype(str)
    .str.strip()
)

# Replace invalid names
invalid_names = [
    "",
    "N/A",
    "NA",
    "None",
    "nan",
    "12345"
]

df.loc[
    df["Customer Name"].isin(invalid_names),
    "Customer Name"
] = "Unknown"


# ============================================================
# Q22. CONVERT ALL NUMERIC COLUMNS
# ============================================================

numeric_columns = [
    "Age",
    "Room Nights",
    "Room Rate",
    "Discount %",
    "Rating",
    "Guests",
    "Cancelled",
    "Gross Amount",
    "Discount Amount",
    "Final Amount"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

print("\nNumeric Data Types:")
print(df[numeric_columns].dtypes)


# ============================================================
# Q23. HANDLE MISSING FINANCIAL VALUES
# ============================================================

financial_columns = [
    "Gross Amount",
    "Discount Amount",
    "Final Amount",
    "Profit"
]

for col in financial_columns:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.extract(r"(-?\d+\.?\d*)")[0]
        )

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )


# Calculate missing Final Amount
df["Final Amount"] = df["Final Amount"].fillna(
    df["Gross Amount"] - df["Discount Amount"]
)

# Calculate missing Profit if possible
if "Profit" in df.columns:
    df["Profit"] = df["Profit"].fillna(
        df["Final Amount"] - df["Gross Amount"]
    )


# ============================================================
# Q24. REMOVE ROWS WITH INVALID DATES
# ============================================================

df["Booking Date"] = pd.to_datetime(
    df["Booking Date"],
    errors="coerce",
    dayfirst=True
)

# Remove rows where date is completely invalid
df = df.dropna(
    subset=["Booking Date"]
)


# ============================================================
# Q25. STANDARDIZE BOOLEAN COLUMNS
# ============================================================

# Is Member = Yes/No
df["Is Member"] = df["Is Member"].replace({
    "1": "Yes",
    "0": "No",
    1: "Yes",
    0: "No"
})

# Cancelled = 0/1
df["Cancelled"] = df["Cancelled"].replace({
    "Yes": 1,
    "No": 0,
    "Y": 1,
    "N": 0
})

df["Cancelled"] = pd.to_numeric(
    df["Cancelled"],
    errors="coerce"
)


# ============================================================
# Q26. TRIM EXTRA SPACES FROM ALL TEXT COLUMNS
# ============================================================

text_columns = df.select_dtypes(
    include=["object"]
).columns

for col in text_columns:
    df[col] = df[col].str.strip()


# ============================================================
# Q27. FIX EXTREME AGE VALUES
# ============================================================

df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Reasonable age range = 18 to 90
df.loc[
    (df["Age"] < 18) | (df["Age"] > 90),
    "Age"
] = np.nan


# ============================================================
# Q28. HANDLE MISSING GENDER VALUES
# ============================================================

df["Gender"] = df["Gender"].replace(
    ["", "nan", "NaN", "None"],
    np.nan
)

df["Gender"] = df["Gender"].fillna(
    "Unknown"
)


# ============================================================
# Q29. CORRECT OBVIOUS TYPOS
# ============================================================

# Room Type
df["Room Type"] = df["Room Type"].replace({
    "delux": "Deluxe",
    "Delux": "Deluxe",
    "exec": "Executive",
    "Exec": "Executive"
})

# City
df["City"] = df["City"].replace({
    "Hydrabad": "Hyderabad",
    "hydrabad": "Hyderabad",
    "Bangalore": "Bengaluru",
    "bangalore": "Bengaluru"
})


# ============================================================
# Q30. HANDLE ROWS WITH TOO MANY MISSING VALUES
# ============================================================

# Calculate percentage of missing values in each row

missing_percentage = df.isnull().mean(axis=1) * 100

# Remove rows with more than 40% missing values

df = df[
    missing_percentage <= 40
]


# ============================================================
# FINAL CHECK
# ============================================================

print("\n========================================")
print("FINAL CLEANED DATASET")
print("========================================")

print("Final Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)


# ============================================================
# SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "C:\\Users\\yuvra\\Downloads\\Hotel_booking_dataset.csv",
    index=False
)

print("\nCleaned dataset saved as:")
print("hotel_booking_cleaned.csv")
```
