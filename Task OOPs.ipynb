```python
# food management

class Food(ABC):
    
    # constructor :
    def __init__(self,f_name,f_price):
        
        self.__Food_name = f_name
        self.__Food_price = f_price
        
    # getter name 
    def get_food_name(self):
        return self.__Food_name
    
    # getter price
    def get_food_price(self):
        return self.__Food_price
    
    # abstract method 
    @abstractmethod
    def prepare(self):
        pass
    
    # abstract method:
    @abstractmethod 
    def deliver(self):
        pass
    
    # display all details 
    def display_all_details(self):
        print(f"Food Name  : {self.__Food_name}")
        print(f"Food Price  : {self.__Food_price}")
        self.prepare()
        self.deliver()
        
    
    
# child class
class Pizza(Food):
    
    def prepare(self):
        print('Pizza is prepared ')
        
    def deliver(self):
        print("Pizza is Delivered")
        
# child class 2:
class Burger(Food):
    
    def prepare(self):
        print('Burger is prepared ')
        
    def deliver(self):
        print("Burger is Delivered") 
        
# child class 3:
class Biryani(Food):
    
    def prepare(self):
        print('Biryani is prepared ')
        
    def deliver(self):
        print("Biryani is Delivered") 
        
# create the class object

pizza = Pizza("Pizza",500)
burger = Burger("Burger",200)
birynai = Biryani("Biryani",700)


birynai.display_all_details()
```

```python
#Task 23/07/2026

from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

# Child Class
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius


# Child Class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Objects
c = Circle(5)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

r = Rectangle(10, 5)
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
```

```python
# Parent  : Employee
# attribute :name , salary,dept all private
# getter and setter : use static method for set_sal :sal > 0
# class atribute : total_emp = 0
# class method : get_total()
# method : work() -- print working meassge
# method calculate_bonus - return of 5% of sal
# method show == print all +bonus

# create the 3 child 
# child 1 : devloper : add langauge, bonus 20 %
# overide work () --writing code 

# child 1 : Designer : add tool, bonus 15 %
# overide work () --Creating UI

# child 1 : manager : add team_size, bonus 20 %
# overide work () --Managing Team

class Employee:
    total_emp = 0  # class attribute
    
    def __init__(self, name, salary, dept):
        # private attributes
        self.__name = name
        self.__salary = salary
        self.__dept = dept
        Employee.total_emp += 1  # increase count
    
    # Getters
    def get_name(self):
        return self.__name
    
    def get_salary(self):
        return self.__salary
    
    def get_dept(self):
        return self.__dept
    
    # Setter with static method
    @staticmethod
    def set_sal(sal):
        if sal > 0:
            return sal
        else:
            print("Salary must be greater than 0")
            return 0
    
    def set_salary(self, sal):
        self.__salary = Employee.set_sal(sal)
    
    @classmethod
    def get_total(cls):
        return f"Total Employees: {cls.total_emp}"
    
    def work(self):
        print("Employee is working")
    
    def calculate_bonus(self):
        return 0.05 * self.__salary  # 5% bonus
      
    def show(self):
        bonus = self.calculate_bonus()
        print(f"Name: {self.__name}")
        print(f"Dept: {self.__dept}")
        print(f"Salary: {self.__salary}")
        print(f"Bonus: {bonus}")
        print("-" * 20)

# Child 1: Developer
class Developer(Employee):
    def __init__(self, name, salary, dept, language):
        super().__init__(name, salary, dept)
        self.language = language
    
    def work(self):  # override
        print("Writing Code")
    
    def calculate_bonus(self):  # override - 20%
        return 0.20 * self.get_salary()
    
    def show(self):
        super().show()
        print(f"Language: {self.language}")

# Child 2: Designer
class Designer(Employee):
    def __init__(self, name, salary, dept, tool):
        super().__init__(name, salary, dept)
        self.tool = tool
    
    def work(self):  # override
        print("Creating UI")
    
    def calculate_bonus(self):  # override - 15%
        return 0.15 * self.get_salary()
    
    def show(self):
        super().show()
        print(f"Tool: {self.tool}")

# Child 3: Manager
class Manager(Employee):
    def __init__(self, name, salary, dept, team_size):
        super().__init__(name, salary, dept)
        self.team_size = team_size
    
    def work(self):  # override
        print("Managing Team")
    
    def calculate_bonus(self):  # override - 20%
        return 0.20 * self.get_salary()
    
    def show(self):
        super().show()D
        print(f"Team Size: {self.team_size}")

# Testing
d1 = Developer("Aarav", 50000, "IT", "Python")
des1 = Designer("Pooja", 40000, "Design", "Figma")
m1 = Manager("Rahul", 80000, "HR", 12)

d1.work()
d1.show()

des1.work()
des1.show()

m1.work()
m1.show()

print(Employee.get_total())
```

# Type 1: Duck Typing

```python
# duck typing 

class CreditCard:    
    def pay(self,amount):
        
        print(f"Paying amount using Credit Card : {amount}")
    
class Debitcard:
    def pay(self,amount):
        print(f"Paying amount using Debit Card : {amount}")

class UPI:
    
    def pay(self,amount):
        print(f"Paying amount using Debit Card : {amount}")
    



def make_payment(payment, amount):
        payment.pay(amount)         # just calls pay() 


# Create objects
c = CreditCard()
d = Debitcard()
upi = UPI()

# Same function — different behavior
make_payment(c,  500)    #  Paying $500 via Credit Card!
make_payment(d,  300)    #  Paying $300 via PayPal!
make_payment(upi, 1000)   #  Paying $1000 via Crypto!
```

```python
class Calculator:
    
    def __init__(self,grade):
        self.Grade = grade 
        
        
        
    # dunder method : __add__
    def __add__(self,other):
        result = self.Grade + other.Grade
        return result
    
    # dunder method : __sub__
    def __sub__(self,other):
        result = self.Grade - other.Grade
        return result
    
    # dunder method : __mult__
    def __mul__(self,other):
        result = self.Grade * other.Grade
        return result
    
    # dunder method : __add__
    def __truediv__(self,other):
        result = self.Grade  / other.Grade
        return result
    
    def __repr__(self):
        return f" Student details Grade = {self.Grade}"
    
    def __str__(self):
        return f" Student details Grade = {self.Grade}"
        
        
# class object 
s1 =Calculator(80)
s2 =Calculator(52)
s3 =Calculator(32)

print(repr(s1))
print(s1) 
```

```python
file = open("details.txt", "r")
content = file.read()
file.close()

words = content.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)
```

```python
file = open("numbers.txt", "w")

for i in range(5):
    num = input("Enter a number: ")
    file.write(num + "\n")

file.close()

file = open("numbers.txt", "r")

total = 0

for line in file:
    total = total + int(line)

file.close()

print("Sum =", total)
```

```python
file = open("student.txt", "r")
content = file.read()
file.close()

old_word = input("Enter word to replace: ")
new_word = input("Enter new word: ")

content = content.replace(old_word, new_word)

file = open("student.txt", "w")
file.write(content)
file.close()

print("Word replaced successfully.")
```

```python
while True:

    print("\n1. Write File")
    print("2. Read File")
    print("3. Append File")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        file = open("data.txt", "w")
        text = input("Enter text: ")
        file.write(text)
        file.close()

    elif choice == 2:
        file = open("data.txt", "r")
        print(file.read())
        file.close()

    elif choice == 3:
        file = open("data.txt", "a")
        text = input("Enter text: ")
        file.write("\n" + text)
        file.close()

    elif choice == 4:
        break

    else:
        print("Invalid choice")
```

```python
file = open("sample.txt", "r")

lines = file.readlines()

file.close()

unique = set(lines)

file = open("unique.txt", "w")

for line in unique:
    file.write(line)

file.close()

print("Duplicate lines removed.")
```
