```python
#With Constructor
class Students:       
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print("Name :", self.name)
        print("Age :", self.age)
        print("Marks :", self.marks)
        print()
        
s1=Students("Sailaja sree", 23, 78)
s2=Students("Ritvika", 20, 85)

s1.display()
s2.display()
```

```python
#Without Constructor
class Student:

    def info(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print(f"Name : {self.name}")
        print(f"Age : {self.age}")
        print(f"Marks : {self.marks}")
        print()

s1 = Student()
s2 = Student()

s1.info("Rahul", 20, 85)
s2.info("Priya", 19, 92)

s1.display()
s2.display()
```

```python
class Students:

    def student_info(self, name, roll_no, class_n):
        self.Name = name
        self.Roll_number = roll_no
        self.Class_name = class_n

    def display_information(self):
        print(f"Student Name : {self.Name}")
        print(f"Student Roll No : {self.Roll_number}")
        print(f"Student Class Name : {self.Class_name}")


# Create object
std1 = Students()
std2 = Students()
# Call first method
std1.student_info("Sumit", 4952, "NareshIT")
std2.student_info("Sumit", 4952, "NareshIT")

# Call second method
std1.display_information()
std2.display_information()
```

```python
class Employee:

    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def increase_salary(self):
        self.salary += 5000
        print("Updated Salary of employee Sailu:", self.salary)


emp = Employee(101, "Sailu", 30000)
emp.increase_salary()
```

```python
class Mobile:

    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def discount(self):
        self.price = self.price - (self.price * 10 / 100)
        print("Price after 10% Discount :", self.price)


m1 = Mobile("Samsung", "A55", 40000)
m1.discount()
```

```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount Deposited :", amount)
        print("Balance :", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount Withdrawn :", amount)
            print("Balance :", self.balance)
        else:
            print("Insufficient Balance")


acc = BankAccount("Rahul", 10000)

acc.deposit(3000)
acc.withdraw(5000)
acc.withdraw(12000)
```

```python
class Car:

    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def increase_speed(self, value):
        if self.speed + value <= 200:
            self.speed += value
            print("Current Speed:", self.speed, "km/h")
        else:
            self.speed = 200
            print("Maximum Speed Reached:", self.speed, "km/h")


car = Car("BMW", 180)

car.increase_speed(10)
car.increase_speed(20)
```

```python
class Temperature:

    def __init__(self, temperature):
        self.temperature = temperature

    def set_temperature(self, value):
        if -20 <= value <= 50:
            self.temperature = value
            print("Temperature Updated:", self.temperature)
        else:
            print("Invalid Temperature")


temp = Temperature(25)

temp.set_temperature(40)
temp.set_temperature(60)
```

```python
class MovieTicket:

    def __init__(self, movie_name, seats):
        self.movie_name = movie_name
        self.seats = seats

    def book_ticket(self, number):
        if number <= self.seats:
            self.seats -= number
            print("Booking Successful")
            print("Remaining Seats:", self.seats)
        else:
            print("Seats Not Available")


movie = MovieTicket("Pushpa 2", 50)

movie.book_ticket(5)
movie.book_ticket(48)
```

```python
class Library:

    def __init__(self, book_name, available_copies):
        self.book_name = book_name
        self.available_copies = available_copies

    def borrow_book(self):
        if self.available_copies > 0:
            self.available_copies -= 1
            print("Book Borrowed")
            print("Available Copies:", self.available_copies)
        else:
            print("Book Not Available")


book = Library("Python", 2)

book.borrow_book()
book.borrow_book()
book.borrow_book()
```

```python
class Product:

    def __init__(self, product_name, quantity):
        self.product_name = product_name
        self.quantity = quantity

    def sell(self, qty):
        if qty <= self.quantity:
            self.quantity -= qty
            print("Product Sold")
            print("Remaining Quantity:", self.quantity)
        else:
            print("Not Enough Stock")


p = Product("Laptop", 10)

p.sell(3)
p.sell(8)
```

```python
class Natural:
    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 15:
            x = self.num
            self.num += 1
            return x
        raise StopIteration

n = Natural()

for i in n:
    print(i)
```
